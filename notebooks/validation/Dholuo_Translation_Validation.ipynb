##Native Speaker Validation of Dholuo Translations for Machine Translation Dataset

#1. Introduction

This notebook presents the native-speaker validation and quality assurance process for a multilingual dataset containing English, Kiswahili, and Dholuo translations. The primary objective of this validation is to ensure that the Dholuo translations are:

Semantically accurate – preserving the meaning of the source sentence.

Grammatically correct – following proper Dholuo grammar rules.

Naturally fluent – sounding like authentic native Dholuo usage.

Orthographically consistent – using correct and consistent spelling.

As a native Dholuo speaker, my role in this project is to perform a linguistic quality review of the translations produced during dataset preparation. Any translation that does not meet the required quality standards is corrected and documented for transparency.

2. Objective

The specific objectives of this notebook are:

Load and inspect the translation dataset.

Identify missing or incomplete review fields.

Validate the Dholuo translations against the English source sentences.

Provide corrected Dholuo translations where necessary.

Document reviewer comments explaining the corrections.

Generate a reviewed dataset for downstream machine translation model training.

#7. Validation Methodology

Each Dholuo translation is reviewed using the following criteria:

Meaning Preservation



The Dholuo translation must convey the same meaning as the English source sentence.

Grammatical Correctness



The sentence must follow correct Dholuo grammar and sentence structure.

Natural Fluency



The translation should sound natural to a native Dholuo speaker rather than a word-for-word translation.

Spelling Consistency



Words should be spelled correctly and consistently throughout the dataset.

#8. Prepare Review Columns

This step ensures that all required review columns exist.

```python
# Ensure review columns exist
required_columns = [
    'Dholuo_accurate',
    'Corrected_Dholuo',
    'Reviewer_comments'
]

for col in required_columns:
    if col not in df.columns:
        df[col] = ''

print('Review columns are ready.')
```

#10. Review

Below is an example of how a native-speaker review is documented.

```python
# Example review for one row
row_index = 0

df.loc[row_index, 'Dholuo_accurate'] = 'No'
df.loc[row_index, 'Corrected_Dholuo'] = 'Luok lweti kinde duto.'
df.loc[row_index, 'Reviewer_comments'] = (
    'Corrected verb form and improved natural fluency.'
)

df.loc[[row_index], ['English', 'Dholuo', 'Corrected_Dholuo',
                     'Dholuo_accurate', 'Reviewer_comments']]
```

#12. Review Progress Tracking

```python
# Count reviewed rows
reviewed = df['Dholuo_accurate'].notna().sum()

# Count approved and corrected
approved = (df['Dholuo_accurate'] == 'Yes').sum()
corrected = (df['Dholuo_accurate'] == 'No').sum()

print(f'Reviewed: {reviewed}')
print(f'Approved: {approved}')
print(f'Corrected: {corrected}')
```
